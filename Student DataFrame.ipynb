```python
import pandas as pd
```

```python
data = {
    'roll': [1,2,3,4,5],
    'name': ['Amit', 'Rhea', 'Sima', 'Raju', 'Babita'],
    'dept': ['CSE', 'IT', None, 'ECE', None],
    'marks': [90, None, 80, 85, 95]
}
```

```python
df=pd.DataFrame(data)
```

```python
print("Student DataFrame:")
print(df)
```

```python
# i. Find out the columns with null values
print("Columns with null values:")
print(df.columns[df.isnull().any()])
```

```python
# ii. Fill the null value of Sima with CSE
df.loc[df['name'] == 'Sima', 'dept'] = 'CSE'
print("After filling Sima's department with CSE:")
print(df)
```

```python
# iii. Delete the null value of Poonita
df = df.dropna(subset=['dept'], how='any')
print("After deleting the null value of Poonita:")
print(df)
```

```python
# iv. Update the department of Raju with VLS1
df.loc[df['name'] == 'Raju', 'dept'] = 'VLS1'
print("After updating Raju's department to VLS1:")
print(df)
```
