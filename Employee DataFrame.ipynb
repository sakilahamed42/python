```python
import pandas as pd
```

```python
data = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Employee_Name': ['Amit', 'Riya', 'Rahul', 'Priya', 'Karan'],
    'Contact_No_1': [9876543210, 9876543211, 9876543212, 9876543213, 9876543214],
    'Project_Location_1': ['Delhi', 'Mumbai', 'Pune', 'Kolkata', 'Chennai'],
    'Contact_No_2': [9123456780, 9123456781, 9123456782, 9123456783, 9123456784],
    'Project_Location_2': ['Mumbai', 'Pune', 'Delhi', 'Delhi', 'Mumbai']
}
```

```python
df = pd.DataFrame(data)
```

```python
print("DataFrame in Wide Format:")
print(df)
```

```python
# Converting Wide Format into Long Format
df_long = pd.melt(
    df,
    id_vars=['Employee_ID', 'Employee_Name'],
    var_name='Details',
    value_name='Value'
)

print("DataFrame in Long Format:")
print(df_long)
```
